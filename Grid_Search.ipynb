```python
import numpy as np
import pandas as pd
df  = pd.read_csv('data_filled_400.csv')
```

```python
data = df.drop(columns=['element2','element1','NO(%)'], axis=1)
target = df['NO(%)']
```

```python
from catboost import CatBoostRegressor
from sklearn.model_selection import GridSearchCV
import numpy as np
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test = train_test_split(data,target,random_state=352,test_size=0.2)  # 数据集


model = CatBoostRegressor(
    verbose=0,  
    early_stopping_rounds=50  
)

# 定义参数网格
param_grid = {
    'depth': [4, 6, 8],
    'learning_rate': [0.01, 0.05, 0.1],
    'l2_leaf_reg': [1, 3, 5],
    'grow_policy': ['SymmetricTree', 'Lossguide'],
    'iterations': [1000],  
}


grid = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=3,  
    scoring='neg_mean_absolute_error',
    n_jobs=-1, 
    verbose=2  
)


grid.fit(x_train, y_train, eval_set=(x_test, y_test))  


print(f"最佳参数: {grid.best_params_}")
print(f"最佳分数: {-grid.best_score_:.4f}")
```
